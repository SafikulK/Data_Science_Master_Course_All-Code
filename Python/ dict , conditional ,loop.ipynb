```python
s = [ 1,4,5,7, 12.25,12.25, "safikul" , "tahe" , True , False,52+25j , 254+85j, (1,2,9), {"name" :"safi" , "num" : 4178} , [ 1,5,8,]]
```

```python
s1_int = []
s2_flote = []
s3_str = []
s4_bool= []
s5_complex = []
s7_dict = []
s8_list = []
s9_Tuple = []
for i in s:
    if type  (i) == int:
        s1_int.append(i)
    elif type (i) == float:
        s2_flote.append(i)
    elif type (i) == str:
        s3_str.append(i)
    elif type  (i) == bool:
        s4_bool.append(i)
    elif type (i) == complex:
        s5_complex.append(i)
    elif type  (i) == dict:
        s7_dict.append(i)
    elif type  (i) == list:
        s8_list.append(i)
    else :
        s9_Tuple.append(i)
        
```
